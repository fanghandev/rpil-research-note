```python
import qmirt
from matplotlib import pyplot as plt
import numpy as np
from pathlib import Path
from plotly import graph_objects as go
```

```python
plot_dir = Path(".")
if not (plot_dir / "geometry_style.json").exists():
    plot_dir = Path("results/cardiac_spect")

phantom_fig = qmirt.plot.wrl.plot_wrl_file(
    plot_dir / "jaszczak_phantom_only.wrl",
    style_sheet=plot_dir / "geometry_style.json",
    title="Jaszczak Phantom Geometry",
)
phantom_fig.show()
```

```python
plot_dir = Path(".")
if not (plot_dir / "geometry_style.json").exists():
    plot_dir = Path("results/cardiac_spect")

combined_fig = qmirt.plot.wrl.plot_wrl_file(
    plot_dir / "scanner_and_phantom_geometry.wrl",
    style_sheet=plot_dir / "geometry_style.json",
    title="Scanner + Jaszczak Phantom Geometry",
)

cube_size = 150.0
half = cube_size / 2.0
cube_vertices = np.array([
    [-half, -half, -half],
    [half, -half, -half],
    [half, half, -half],
    [-half, half, -half],
    [-half, -half, half],
    [half, -half, half],
    [half, half, half],
    [-half, half, half],
])

cube_triangles = np.array([
    [0, 1, 2], [0, 2, 3],
    [4, 5, 6], [4, 6, 7],
    [0, 1, 5], [0, 5, 4],
    [1, 2, 6], [1, 6, 5],
    [2, 3, 7], [2, 7, 6],
    [3, 0, 4], [3, 4, 7],
])

# combined_fig.add_trace(
#     go.Mesh3d(
#         x=cube_vertices[:, 0],
#         y=cube_vertices[:, 1],
#         z=cube_vertices[:, 2],
#         i=cube_triangles[:, 0],
#         j=cube_triangles[:, 1],
#         k=cube_triangles[:, 2],
#         color="green",
#         opacity=0.05,
#         name="FOV fill",
#         legendgroup="FOV",
#         showlegend=True,
#         flatshading=True,
#         hoverinfo="skip",
#     )
# )

cube_edges = [
    (0, 1), (1, 2), (2, 3), (3, 0),
    (4, 5), (5, 6), (6, 7), (7, 4),
    (0, 4), (1, 5), (2, 6), (3, 7),
]

edge_x = []
edge_y = []
edge_z = []
for start, end in cube_edges:
    edge_x.extend([cube_vertices[start, 0], cube_vertices[end, 0], None])
    edge_y.extend([cube_vertices[start, 1], cube_vertices[end, 1], None])
    edge_z.extend([cube_vertices[start, 2], cube_vertices[end, 2], None])

combined_fig.add_trace(
    go.Scatter3d(
        x=edge_x,
        y=edge_y,
        z=edge_z,
        mode="lines",
        name="FOV wireframe",
        legendgroup="FOV",
        showlegend=True,
        line=dict(color="green", width=2),
        hoverinfo="skip",
    )
)

combined_fig.update_layout(scene=dict(aspectmode="data"))
combined_fig.show()
```
